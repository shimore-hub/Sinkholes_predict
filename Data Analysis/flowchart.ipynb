```python
from graphviz import Digraph

# Initialize the flowchart
flowchart = Digraph(format="png", comment="Workflow for Sinkhole Mapping")
flowchart.attr(rankdir='TB', size='8')

# General Steps
flowchart.node("A", "Data Acquisition")
flowchart.node("B", "Create Mask Labels\nfrom Polygons")
flowchart.node("C", "Normalize Data")
flowchart.node("D", "Divide Data into\nTrain and Test Groups")
flowchart.node("E", "Train Models")
flowchart.node("F", "Test Models")
flowchart.node("G", "Evaluate Results\nUsing Metrics")

# Logistic Regression-Specific Steps
flowchart.node("LR1", "Create Morphometric\nIndices", style="filled", color="orange")
flowchart.node("LR2", "Check Collinearity and\nRemove Indices", style="filled", color="orange")

# Arrows for Workflow
flowchart.edge("A", "B")
flowchart.edge("B", "LR1", label="(LR Only)", style="dashed", color="orange")
flowchart.edge("LR1", "LR2", style="dashed", color="orange")
flowchart.edge("LR2", "C", style="dashed", color="orange")
flowchart.edge("B", "C")
flowchart.edge("C", "D")
flowchart.edge("D", "E")
flowchart.edge("E", "F")
flowchart.edge("F", "G")

# Render the flowchart
flowchart.render("workflow_sinkhole_mapping", cleanup=True)
print("Flowchart created and saved as 'workflow_sinkhole_mapping.png'")
```

```python
from graphviz import Digraph

# Initialize the flowchart
flowchart = Digraph(format="png", comment="Workflow for Sinkhole Mapping")
flowchart.attr(rankdir='TB', size='8')

# Add main groups (subgraphs in Graphviz)
with flowchart.subgraph(name='cluster_data_preparation') as data_prep:
    data_prep.attr(label="Data Preparation", style="filled", color="lightblue")
    data_prep.node("A", "Data Acquisition")
    data_prep.node("B", "Create Mask Labels\nfrom Polygons")
    data_prep.node("LR1", "Create Morphometric\nIndices (LR Only)", style="filled", color="orange")
    data_prep.node("LR2", "Check Collinearity\nand Remove Indices (LR Only)", style="filled", color="orange")
    data_prep.node("C", "Normalize Data")
    data_prep.node("D", "Divide Data into\nTrain and Test Groups")
    data_prep.edges([("A", "B"), ("B", "LR1"), ("LR1", "LR2"), ("LR2", "C"), ("B", "C"), ("C", "D")])

with flowchart.subgraph(name='cluster_model_training') as training:
    training.attr(label="Model Training", style="filled", color="lightgreen")
    training.node("E", "Train Models")

with flowchart.subgraph(name='cluster_testing_evaluation') as testing:
    testing.attr(label="Model Testing & Evaluation", style="filled", color="lightyellow")
    testing.node("F", "Test Models")
    testing.node("G", "Evaluate Results\nUsing Metrics")
    testing.edge("F", "G")

# Connect the main groups
flowchart.edge("D", "E")
flowchart.edge("E", "F")

# Render the flowchart
flowchart.render("workflow_sinkhole_mapping_grouped", cleanup=True)
print("Flowchart created and saved as 'workflow_sinkhole_mapping_grouped.png'")
```
